1. It is HashableBecause it cannot change, a frozenset is hashable. This means, unlike a regular set, you can use a frozenset as:A key in a dictionary.An element inside another set (creating a set of sets).

```python
# ❌ A regular set as a key will raise a TypeError
# my_dict = {{1, 2}: "values"}

#  A frozenset works perfectly
valid_dict = {frozenset([1, 2]): "values"}
print(valid_dict[frozenset([1, 2])])  # Output: "values"
```

[ L ] Local       --> Inside the current function
  [ E ] Enclosing   --> Inside any enclosing/nested functions
    [ G ] Global      --> At the top level of the script/module
      [ B ] Built-in  --> Pre-defined Python names (like print, len)

 Warning: Modifying the dictionary returned by locals() inside a function does not change the actual local variables. Python optimizes local variable lookups, making this dictionary read-only.

```python
def my_func():
    a = 1
    b = "hello"
    locals()['a']=10
    print(locals())  # Prints: {'a': 1, 'b': 'hello'}

my_func()
print(locals()==globals())
```

3. vars()Without arguments, vars() behaves exactly like locals(). With an argument (like a class, module, or object instance), it returns that object's internal namespace dictionary (__dict__).

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Alice", 30)

# Extract object properties as a clean dictionary
print(vars(p))  # Prints: {'name': 'Alice', 'age': 30}

# Modifying vars() modifies the object directly
vars(p)['age'] = 31
print(p.age)    # Prints: 31
```

================packing && unpacking=======================

Packing and unpacking in Python are operations that allow you to group multiple values together into a single collection, or extract individual elements from a collection back into separate variables

```python
packed_list = [1,3,5,2,4,3]
packed_tuple = 1,4,5,6,2,3
packed_set = {1,2,2,3,4,5,4}
packed_dict = {'a':1,'b':2,'c':3}
print(packed_list)
print(packed_tuple)
print(packed_set)
print(packed_dict)

print("==============Unpacking==============")
a,b,c,d,e,f = packed_list
print(a,b,c,d,e,f)
a,*b,c = packed_tuple
print(a,b,c)
a,*b,c = packed_set
print(a,b,c)
a,*b,c = packed_dict
print(a,b,c)
```

Extended Unpacking with * (The Asterisk Operator)If you do not know how many elements are in a collection, or you only care about specific parts, use * to capture the remaining items into a new list

```python
# Capture the remainder at the end
scores = [95, 88, 72, 64, 50]
highest, *others = scores
print(highest)  # Output: 95
print(others)   # Output: [88, 72, 64, 50]

# Capture the remainder in the middle
items = ["A", 1, 2, 3, "Z"]
start, *middle, end = items
print(middle)   # Output: [1, 2, 3]

# Ignore unwanted values using '_' (convention for throwaway variables)
data = ("John", "Doe", 45, "Manager", "New York")
first_name, last_name, *_, location = data
print(location) # Output: New York
```

Unpacking with the double asterisk (**) is used exclusively with dictionaries. It extracts the key-value pairs from a dictionary and spreads them out as named keyword arguments


Unpacking Dictionary into a Function CallThe most common use case is passing a dictionary of settings or data directly into a function that expects matching keyword arguments

```python
def create_user(username, email, role):
    print(f"User: {username} | Email: {email} | Role: {role}")

user_data = {
    "username": "coder123",
    "email": "coder@mail.com",
    "role": "Admin"
}

# Unpacking the dictionary directly into the function
create_user(**user_data)
# Equivalent to: create_user(username="coder123", email="coder@mail.com", role="Admin")
```

In Python, everything is a first-class object (even with a quick typo in the name!). This means that things traditionally treated as rigid syntax in other languages—such as functions, classes, modules, and data types—enjoy the exact same privileges as raw data like integers or strings

4 pillars of first class object:
1. Stored in a Variable or Data Structure
2. Passed as an Argument to a Function
3. Returned as a Result from a Function
4. Possess Attributes and Methods

```python
# 1 stored in a variable
def greet(name):
    return f"Hello, {name}!"
var = greet
print(var("sumit"))

# 2   passed as an argument
def apply(func, x):
    return func(x)
print(apply(var, "Bob"))

# 3 returning a function from another function

def outer_func():
  def inner_func():
    return ("hello ! from inner function")
  return inner_func

inner = outer_func()
print(inner())

# 4 Possess Attributes and Methods
print(inner.__name__)
print(inner())

def process_data():
    pass

# Dynamically attaching custom attributes to a function object
process_data.version = "1.0.4"
process_data.author = "Admin"

print(process_data.version)  # Output: 1.0.4
print(process_data.__name__) # Output: process_data (Built-in attribute)


```

=============filter()===========

The filter() function is a built-in Python higher-order function that takes an iterable (like a list, tuple, or set) and extracts only the elements that meet a specific condition.It evaluates each item using a function that returns True or False. If the function returns True, the element is kept; otherwise, it is dropped.


filter(function, iterable)

function: A function that returns a boolean (True or False) for each element.

iterable: The collection of items you want to filter.

```python
def is_even(num):
    return num % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

# filter() returns a lazy filter object iterator
result = filter(is_even, numbers)

# Convert the iterator to a list to read the values
print(list(result))  # Output: [2, 4, 6]
```

=============================REduce() ==========================


The reduce() function is a higher-order function that continually applies a specific mathematical or logical operation to pairings of elements in an iterable, boiling the entire collection down to a single cumulative value.

Unlike map() and filter(), reduce() is not a built-in function in Python. Because it can be tricky to read, it was moved to the functools module.

```python
from functools import reduce

numbers = [1, 2, 3, 4]

# x is the running total, y is the next element in the list
total_sum = reduce(lambda x, y: x + y, numbers)
print(total_sum)  # Output: 10
```

=================Closure =================================================

A Python closure is an inner function object that retains access to variables from its outer (enclosing) scope, even after the outer function has finished executing. It essentially allows a nested function to "remember" the environment in which it was created

3 Conditions for a ClosureFor a closure to exist, three specific conditions must be met:

1. Nested Function: A function must be defined inside another function.

2. Enclosing Scope Reference: The inner function must reference a variable from the outer function.

3. Returned Function: The outer function must return the inner function object

```python
def outer_func():
  var =200
  def inner_func():
    print('from ineer function accessing enclosed variable: ',var)
  return inner_func

fn = outer_func()
fn()
```

```python
def make_multiplier(factor):
    # Enclosing scope variable: factor
    def multiplier(number):
        return number * factor  # Inner function accesses 'factor'

    return multiplier  # Returns the inner function object

# Create closure instances
double = make_multiplier(2)
triple = make_multiplier(3)

print('double i pointing to --->',double.__name__,double(5))  # Output: 10  ---yaha pe double jo ki inner function hai encloser variable: factor ko yaad rakhta hai
print(triple(5))  # Output: 15

print(double.__closure__[0].cell_contents)  # Output: 2
print(triple.__closure__[0].cell_contents)  # Output: 2


```

====================================**DEcorator**=====================

A decorator in Python is a design pattern that allows you to modify or extend the behavior of a function or class without permanently changing its source code.


Decorators wrap another function, dynamically running code before and after the wrapped function executes. Because they rely entirely on closures, they can remember and manipulate data across function calls.

```python
def my_decorator(func):
  def wrapper():
    print('something is happening before the function is called')
    func()
    print('something is happening after the function is called')
  return wrapper

@my_decorator
def say_hello():
  print('hello, World')

@my_decorator
def say_hi():
  print('hi.........., ')

@my_decorator
def say_bye():
  print('bye bye, World')

#fn = my_decorator(say_hello)
#fn()
say_hello()
say_hi()
say_bye()
```

```python
def log_args(func):
    def wrapper(*args, **kwargs):
        print(func.__name__, f"was called with arguments: {args}")
        return func(*args, **kwargs)  # Captures and returns the original output
    return wrapper

@log_args
def add_numbers(a, b):
    return a + b

print(add_numbers(5, 10))
```

```python
import time
```

```python
def cal_tim(func):
  def wrapper(*args,**kwargs):
    print(f'execution has started for function: {func.__name__}')
    start_tim = time.perf_counter()
    res =  func(*args,**kwargs)
    end_tim = time.perf_counter()
    print(f'time taken by function: {func.__name__} is = {end_tim-start_tim:.6f}')
    return res
  return wrapper

def slow_func():
  time.sleep(2)
  print('this is from ....slow function')

fn = cal_tim(slow_func)
fn()
@cal_tim
def cub(numbers):
  l = []
  for i in numbers:
    l.append(i**3)
  return l
@cal_tim
def square(numbers):
  l = []
  for i in numbers:
    l.append(i**2)
  return l

cub(range(1,10))

square(range(1,10))


```

**Decorator chaining** means applying more than one decorator to a single function.When you chain decorators, Python applies them from bottom to top (innermost to outermost) at definition time, but they execute like layers of an onion from top to bottom (outermost to innermost) when the function is called.

```python
def bold(func):
    def wrapper(*args, **kwargs):
        print("-> Bold wrapper START")
        result = func(*args, **kwargs)
        print("-> Bold wrapper END")
        return f"<b>{result}</b>"
    return wrapper

def italic(func):
    def wrapper(*args, **kwargs):
        print("  -> Italic wrapper START")
        result = func(*args, **kwargs)
        print("  -> Italic wrapper END")
        return f"<i>{result}</i>"
    return wrapper

@bold
@italic
def greet(name):
    print("    => Core greet() function executing")
    return f"Hello, {name}"

print("--- Calling greet() ---")
final_output = greet("Alice")
print("\nFinal Output:", final_output)
```

Why did it execute in this order?

greet = bold(italic(greet)): The italic decorator wraps the original greet. Then, the bold decorator wraps that entire package.

Going In: When you run greet("Alice"), you trigger the outermost layer first (bold). It prints its start message, then hands off execution to the inner function it wrapped (italic). italic prints its start message and triggers the core function.

Coming Out: The core function returns its string. italic modifies it to <i>Hello, Alice</i> and returns it. bold receives that result, modifies it to <b><i>Hello, Alice</i></b>, and completes the cycle.

===================JSON====================


```python
import json
import os

# 1. Python Dictionary (Python Object)
data = {
    "name": "Alice",
    "age": 30,
    "isStudent": False,
    "courses": [{"title": "History 101", "credits": 3}, {"title": "Math 200", "credits": 4}],
    "contact": {"email": "alice@example.com", "phone": None}
}

print("Original Python data:")
print(data)
print(f"Type of original data: {type(data)}\n")

# 2. Encoding: Python object to JSON string (json.dumps)
json_string = json.dumps(data, indent=4) # indent makes it human-readable

print("Encoded JSON string:")
print(json_string)
print(f"Type of JSON string: {type(json_string)}\n")

# 3. Decoding: JSON string to Python object (json.loads)
decoded_data = json.loads(json_string)

print("Decoded Python data from string:")
print(decoded_data)
print(f"Type of decoded data: {type(decoded_data)}\n")

# 4. Encoding: Python object to JSON file (json.dump)
file_name = "data.json"
with open(file_name, 'w') as json_file:
    json.dump(data, json_file, indent=4)

print(f"Python data successfully written to {file_name}\n")

# 5. Decoding: JSON file to Python object (json.load)
loaded_data = {}
with open(file_name, 'r') as json_file:
    loaded_data = json.load(json_file)

print("Loaded Python data from file:")
print(loaded_data)
print(f"Type of loaded data: {type(loaded_data)}\n")

# Clean up the created file
os.remove(file_name)
print(f"Cleaned up: {file_name} removed.")
```

===================AST==========

```python
import ast

# Define a simple Python code string
code_string = """
def add(a, b):
    result = a + b
    return result
"""

print("Original Python code:\n" + code_string)

# 1. Parse the code string into an AST
tree = ast.parse(code_string)

print("\n--- AST Structure (using ast.dump) ---")
# ast.dump provides a compact, single-line representation
print(ast.dump(tree, indent=4))

# 2. Traverse the AST (using ast.walk for simplicity)
print("\n--- Traversing AST nodes (using ast.walk) ---")
print("Node type: Node fields")
for node in ast.walk(tree):
    # Print the type of the node and its fields
    # Exclude internal fields like lineno, col_offset, end_lineno, end_col_offset for brevity
    node_fields = {k: getattr(node, k) for k in node._fields if not k.startswith(('lineno', 'col_offset', 'end_lineno', 'end_col_offset'))}
    print(f"{type(node).__name__}: {node_fields}")
```

==================Regrex======================

```python
import re

# Sample text to work with
text = "The quick brown fox jumps over the lazy dog. The fox is agile and quick."

print("Original Text:", text)
print("---------------------------------------------------\n")

# 1. re.search(): Finds the first occurrence of a pattern in a string
pattern1 = r"fox"
match1 = re.search(pattern1, text)
if match1:
    print(f"re.search('{pattern1}', text): Found '{match1.group()}' at index {match1.start()}-{match1.end()}")
else:
    print(f"re.search('{pattern1}', text): No match found.")

pattern_digits = r"\d+" # Matches one or more digits
match_digits = re.search(pattern_digits, "There are 123 apples.")
if match_digits:
    print(f"re.search('{pattern_digits}', 'There are 123 apples.'): Found '{match_digits.group()}'")

print("---------------------------------------------------\n")

# 2. re.match(): Checks for a match only at the beginning of the string
pattern2 = r"The"
match2 = re.match(pattern2, text)
if match2:
    print(f"re.match('{pattern2}', text): Found '{match2.group()}' at beginning.")

pattern2_fail = r"fox"
match2_fail = re.match(pattern2_fail, text)
if not match2_fail:
    print(f"re.match('{pattern2_fail}', text): No match at beginning.")

print("---------------------------------------------------\n")

# 3. re.findall(): Finds all non-overlapping occurrences of a pattern
pattern3 = r"quick"
all_matches3 = re.findall(pattern3, text)
print(f"re.findall('{pattern3}', text): {all_matches3}")

pattern3_words = r"\b\w+\b" # Matches whole words
all_words = re.findall(pattern3_words, text)
print(f"re.findall('{pattern3_words}', text) (first 10 words): {all_words[:10]}...")

print("---------------------------------------------------\n")

# 4. re.sub(): Replaces occurrences of a pattern
pattern4 = r"fox"
replacement4 = "cat"
new_text = re.sub(pattern4, replacement4, text)
print(f"re.sub('{pattern4}', '{replacement4}', text):\n{new_text}")

pattern_spaces = r"\s+" # Matches one or more whitespace characters
new_text_compressed = re.sub(pattern_spaces, "_", text)
print(f"re.sub('{pattern_spaces}', '_', text):\n{new_text_compressed}")

print("---------------------------------------------------\n")

# 5. Using flags (e.g., re.IGNORECASE)
pattern_case = r"the"
case_insensitive_matches = re.findall(pattern_case, text, re.IGNORECASE)
print(f"re.findall('{pattern_case}', text, re.IGNORECASE): {case_insensitive_matches}")

print("---------------------------------------------------\n")

# 6. Capturing groups
pattern_groups = r"(\w+) (brown|lazy) (\w+)"
match_groups = re.search(pattern_groups, text)
if match_groups:
    print(f"Capturing groups for '{pattern_groups}':")
    print(f"  Full match: {match_groups.group(0)}")
    print(f"  Group 1: {match_groups.group(1)}")
    print(f"  Group 2: {match_groups.group(2)}")
    print(f"  Group 3: {match_groups.group(3)}")
    print(f"  All groups: {match_groups.groups()}")
```
